## Practical 1: Python Arithmetic Operations

**Aim:** To write a Python program to accept two numbers and display their sum, difference, product, and division.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program takes two numbers from the user and performs addition, subtraction, multiplication, and division. It displays the results of all four operations.

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Sum =", a + b)
print("Difference =", a - b)
print("Product =", a * b)
if b != 0:
    print("Division =", a / b)
else:
    print("Division is not possible")
```

## Practical 3: Area of a Circle

**Aim:** To write a Python program to calculate the area of a circle using the radius entered by the user.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts the radius from the user and calculates the area of the circle using the formula π × radius × radius.

```python
radius = float(input("Enter radius: "))
area = 3.14 * radius ** 2
print("Area of circle =", area)
```
