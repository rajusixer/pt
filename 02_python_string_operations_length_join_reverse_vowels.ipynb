## Practical 5: String Length and Characters

**Aim:** To write a Python program to accept a string and display its length, first character, and last character.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a string and uses string operations to display its length, first character, and last character.

```python
text = input("Enter a string: ")

print("Length =", len(text))

if len(text) > 0:
    print("First character =", text[0])
    print("Last character =", text[-1])
else:
    print("String is empty")
```

## Practical 6: Join Two Strings

**Aim:** To write a Python program to accept two strings and join them to form a single string.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts two strings and joins them using the string concatenation operator to form one string.

```python
a = input("Enter first string: ")
b = input("Enter second string: ")

result = a + " " + b

print("Combined string =", result)
```

## Practical 7: Reverse a String

**Aim:** To write a Python program to accept a string and display it in reverse order.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a string and displays its characters in reverse order.

```python
text = input("Enter a string: ")

reverse = ""
for ch in text:
    reverse = ch + reverse

print("Reversed string =", reverse)
```

## Practical 8: Count Vowels

**Aim:** To write a Python program to count the number of vowels in a given string.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program checks the characters of a string and counts the vowels: a, e, i, o, and u.

```python
text = input("Enter a string: ")

count = 0
for ch in text:
    if ch in "AEIOUaeiou":
        count = count + 1

print("Total vowels =", count)
```
