## Practical 9: Positive, Negative, or Zero

**Aim:** To write a Python program to check whether a given number is positive, negative, or zero.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a number and uses conditional statements to identify whether it is positive, negative, or equal to zero.

```python
num = float(input("Enter a number: "))

if num > 0:
    print("Number is positive")
elif num < 0:
    print("Number is negative")
else:
    print("Number is zero")
```

## Practical 10: Even or Odd

**Aim:** To write a Python program to check whether a given number is even or odd.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses the modulus operator to check whether a number is divisible by 2 and displays whether it is even or odd.

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Number is even")
else:
    print("Number is odd")
```

## Practical 11: Find the Larger Number

**Aim:** To write a Python program to accept two numbers and display the larger number using an if-else statement.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program compares two input numbers using an if-else statement and displays the larger number.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("Larger number =", a)
elif b > a:
    print("Larger number =", b)
else:
    print("Both numbers are equal")
```

## Practical 12: Pass or Fail

**Aim:** To write a Python program to accept marks and display “Pass” if the marks are 40 or above; otherwise display “Fail”.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts marks and uses a condition to display “Pass” when the marks are 40 or above, and “Fail” otherwise.

```python
marks = int(input("Enter marks: "))

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```
