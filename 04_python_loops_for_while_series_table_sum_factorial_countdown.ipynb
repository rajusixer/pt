## Practical 13: Print Numbers Using a For Loop

**Aim:** To write a Python program to accept a number and print numbers from 1 to that number using a for loop.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a number and uses a for loop to print all integers from 1 up to the entered number.

```python
n = int(input("Enter a number: "))

for i in range(1, n + 1):
    print(i)
```

## Practical 14: Even Numbers from 1 to 50

**Aim:** To write a Python program to print all even numbers from 1 to 50.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses a loop and a condition to display all even numbers between 1 and 50.

```python
for i in range(1, 51):
    if i % 2 == 0:
        print(i)
```

## Practical 15: Multiplication Table

**Aim:** To write a Python program to print the multiplication table of a number entered by the user.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a number and uses a loop to display its multiplication table.

```python
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

## Practical 16: Sum from 1 to n

**Aim:** To write a Python program to calculate the sum of numbers from 1 to n using a loop.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program accepts a number n and uses a loop to add the integers from 1 through n, then displays the total.

```python
n = int(input("Enter a number: "))

total = 0
for i in range(1, n + 1):
    total = total + i

print("Sum =", total)
```

## Practical 17: Factorial Using a For Loop

**Aim:** To write a Python program to calculate the factorial of a number using a for loop.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program calculates the factorial by multiplying the positive integers from 1 up to the entered number using a for loop.

```python
n = int(input("Enter a number: "))

fact = 1
for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
```

## Practical 18: Countdown Using a While Loop

**Aim:** To write a Python program to print numbers from 10 to 1 using a while loop.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses a while loop to print numbers in descending order from 10 to 1.

```python
i = 10

while i >= 1:
    print(i, end=" ")
    i -= 1
```
