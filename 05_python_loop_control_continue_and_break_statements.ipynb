## Practical 19: Continue Statement

**Aim:** To write a Python program to skip the number 5 while printing numbers from 1 to 10 using the continue statement.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses the continue statement to skip 5 and print the other numbers from 1 to 10.

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

## Practical 20: Break Statement

**Aim:** To write a Python program to stop printing numbers when the value reaches 7 using the break statement.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses the break statement to stop a loop when the number reaches 7.

```python
for i in range(1, 11):
    if i == 7:
        break
    print(i)
```
