## Practical 21: Math Module

**Aim:** To write a Python program using the math module to find the square root and power of a given number.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program uses Python's math module to calculate and display the square root and power of a given number.

```python
import math

n = float(input("Enter a number: "))
p = float(input("Enter power: "))

if n >= 0:
    print("Square root =", math.sqrt(n))
else:
    print("Square root is not defined for a negative number")
print("Power =", math.pow(n, p))
```

## Practical 22: Function to Add Two Numbers

**Aim:** To write a custom function that accepts two numbers and returns their sum.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program defines a function that takes two numbers as arguments, adds them, and returns the result.

```python
def add(a, b):
    return a + b

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print("Sum =", add(x, y))
```

## Practical 23: Area of a Rectangle

**Aim:** To write a custom function to calculate the area of a rectangle.

**Objective:** To understand and apply the Python concept required to complete this practical.

**Theory:** This program defines a function that accepts the length and breadth of a rectangle and returns its area.

```python
def area(length, breadth):
    return length * breadth

l = float(input("Enter length: "))
b = float(input("Enter breadth: "))

print("Area of rectangle =", area(l, b))
```
