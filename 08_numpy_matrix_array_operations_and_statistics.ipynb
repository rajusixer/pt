## Program 1: 4 x 4 Matrix - Diagonal Elements and Trace

**Aim:** To write a NumPy program to create a 4 x 4 matrix and display its diagonal elements and trace.

**Objective:** To create a 2-D array (matrix) using NumPy; to extract the main diagonal using numpy.diagonal(); to compute the trace (sum of diagonal elements) using numpy.trace().

**Theory:** NumPy (Numerical Python) is a library for fast array computation. Its core object is the ndarray, an N-dimensional homogeneous array.

The diagonal of a square matrix consists of the elements A[i][i]. `np.diagonal(A)` returns these elements.

The trace of a square matrix is the sum of its main diagonal elements: trace(A) = A[0][0] + A[1][1] + ... + A[n-1][n-1]. NumPy provides `np.trace(A)` for this.

```python
import numpy as np

matrix = np.arange(1, 17).reshape(4, 4)
print("Matrix:")
print(matrix)

diag = np.diagonal(matrix)
print("\nDiagonal elements:", diag)

trace = np.trace(matrix)
print("Trace of the matrix:", trace)
```

## Program 2: Concatenate Two Arrays

**Aim:** To write a NumPy program to concatenate two arrays into a single array.

**Objective:** To create two NumPy arrays; to join them using numpy.concatenate(); to understand concatenation along different axes.

**Theory:** Concatenation means joining two or more arrays into one. `np.concatenate((a, b), axis=0)` joins arrays along an existing axis.

For 1-D arrays there is only one axis, so the result is a longer 1-D array. For 2-D arrays, axis=0 stacks rows (vertical) and axis=1 joins columns (horizontal).

The arrays must have the same shape in all dimensions except the one along which they are joined.

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([50, 60, 70])
print("Array 1:", a)
print("Array 2:", b)

result = np.concatenate((a, b))
print("Concatenated array:", result)

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])

print("\nAxis 0 (rows):")
print(np.concatenate((x, y), axis=0))
print("Axis 1 (columns):")
print(np.concatenate((x, y), axis=1))
```

## Program 3: Replace Elements Greater Than 50 with 100

**Aim:** To write a NumPy program to replace all elements greater than 50 with 100 in a given array.

**Objective:** To use boolean (conditional) indexing in NumPy; to modify array elements that satisfy a condition.

**Theory:** Boolean indexing lets us select array elements using a condition. The expression `arr > 50` returns a boolean array of True/False values.

Using this boolean mask as an index, `arr[arr > 50] = 100` assigns 100 only to the positions where the condition is True. This is vectorised and much faster than a Python loop.

Alternatively, `np.where(condition, x, y)` returns x where the condition is True and y otherwise.

```python
import numpy as np

arr = np.array([12, 55, 78, 34, 90, 50, 23, 67, 45, 101])
print("Original array:", arr)

arr[arr > 50] = 100
print("Modified array:", arr)
```

## Program 4: Median, Variance and Standard Deviation

**Aim:** To write a NumPy program to find the median, variance, and standard deviation of an array.

**Objective:** To use np.median(), np.var() and np.std(); to understand measures of central tendency and dispersion.

**Theory:** Median is the middle value of sorted data (average of two middle values if the count is even); it is robust to outliers.

Variance measures the average squared deviation from the mean: Var = sum((x - mean)^2) / n.

Standard deviation is the square root of variance. It is in the same unit as the data. NumPy uses the population formula (ddof=0) by default.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])
print("Array:", data)
print("Median            :", np.median(data))
print("Variance          :", np.var(data))
print("Standard deviation:", np.std(data))
```

## Program 5: Unique Elements and Their Frequencies

**Aim:** To write a NumPy program to find the unique elements and their frequencies in an array.

**Objective:** To use np.unique() with return_counts=True; to count occurrences of each distinct value.

**Theory:** `np.unique(arr)` returns the sorted distinct values of an array.

When `return_counts=True` is passed, it additionally returns an array with the number of times each unique value appears, i.e. its frequency.

```python
import numpy as np

arr = np.array([1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5])
print("Array:", arr)

values, counts = np.unique(arr, return_counts=True)
print("\nUnique elements:", values)
print("Frequencies    :", counts)
print()
for v, c in zip(values, counts):
    print(f"{v} -> {c} time(s)")
```
