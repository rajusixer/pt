## Program 6: Add a New Row to a DataFrame

**Aim:** To write a Pandas program to add a new row to an existing student DataFrame.

**Objective:** To create a DataFrame; to append a new record using loc and pd.concat().

**Theory:** Pandas is a data analysis library; its main structure, the DataFrame, is a 2-D labelled table of rows and columns.

A new row can be added in two ways: (i) `df.loc[len(df)] = [values]` assigns a row at a new index, or (ii) `pd.concat([df, new_df], ignore_index=True)` joins another DataFrame below. `DataFrame.append()` was removed in Pandas 2.0.

```python
import pandas as pd

df = pd.DataFrame({
    "Roll": [1, 2, 3, 4],
    "Name": ["Amit", "Riya", "Karan", "Neha"],
    "Department": ["CSE", "IT", "CSE", "ECE"],
    "Marks": [78, 85, 62, 91]
})
print("Original DataFrame:")
print(df)

df.loc[len(df)] = [5, "Sanjay", "IT", 74]
print("\nAfter adding a row using loc:")
print(df)

new_row = pd.DataFrame({"Roll": [6], "Name": ["Pooja"], "Department": ["ECE"], "Marks": [88]})
df = pd.concat([df, new_row], ignore_index=True)
print("\nAfter adding a row using concat:")
print(df)
```

## Program 7: Identify and Display Missing Values

**Aim:** To write a Pandas program to identify and display missing values in a DataFrame.

**Objective:** To detect missing (NaN) values using isnull() / isna(); to count missing values per column and display rows containing them.

**Theory:** Real-world data often has missing values, represented in Pandas as NaN (Not a Number) or None.

`df.isnull()` returns a boolean DataFrame (True where a value is missing). Applying `.sum()` gives the count of missing values for each column. `df[df.isnull().any(axis=1)]` displays only the rows that contain at least one missing value.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan", "Neha", "Sanjay"],
    "Department": ["CSE", None, "CSE", "ECE", "IT"],
    "Marks": [78, 85, np.nan, 91, np.nan]
})
print("DataFrame:")
print(df)

print("\nBoolean mask of missing values:")
print(df.isnull())

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nRows containing missing values:")
print(df[df.isnull().any(axis=1)])
```

## Program 8: Remove Duplicate Rows

**Aim:** To write a Pandas program to remove duplicate rows from a DataFrame.

**Objective:** To detect duplicate rows using duplicated(); to remove them using drop_duplicates().

**Theory:** Duplicate rows are records that are identical to an earlier row. They can distort counts and averages.

`df.duplicated()` returns True for each repeated row. `df.drop_duplicates()` removes them, keeping the first occurrence by default (keep='first'). The `subset` parameter restricts the comparison to chosen columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Amit", "Neha", "Riya"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"],
    "Marks": [78, 85, 78, 91, 85]
})
print("Original DataFrame:")
print(df)

print("\nDuplicate flags:")
print(df.duplicated())

clean = df.drop_duplicates().reset_index(drop=True)
print("\nAfter removing duplicates:")
print(clean)
```

## Program 9: Group by Department and Average Marks

**Aim:** To write a Pandas program to group student records by department and calculate the average marks for each department using groupby().

**Objective:** To understand the split-apply-combine approach; to use groupby() with the mean() aggregation.

**Theory:** `groupby()` follows the split-apply-combine strategy: (1) split data into groups based on a key column, (2) apply a function (sum, mean, count, etc.) to each group, (3) combine the results into a new object.

`df.groupby('Department')['Marks'].mean()` groups rows by department and returns the average marks of each group.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan", "Neha", "Sanjay", "Pooja", "Rahul", "Simran"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "ECE", "CSE", "IT"],
    "Marks": [78, 85, 62, 91, 74, 88, 70, 80]
})
print("Student records:")
print(df)

avg = df.groupby("Department")["Marks"].mean()
print("\nAverage marks by department:")
print(avg)
```

## Program 10: Assign Grades Using apply()

**Aim:** To write a Pandas program to apply a function to a DataFrame column to assign grades (A, B, C, or D) based on students' marks.

**Objective:** To define a user-defined grading function; to apply it to a column using Series.apply() and store the result in a new column.

**Theory:** The `apply()` method runs a function on every element of a column (Series) and returns a new Series.

Grading rule used: marks >= 80 -> A, 60 to 79 -> B, 40 to 59 -> C, below 40 -> D. The function is written once and applied to all rows, creating a new column 'Grade'.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan", "Neha", "Sanjay", "Pooja"],
    "Marks": [78, 85, 62, 91, 35, 48]
})

def assign_grade(m):
    if m >= 80:
        return "A"
    elif m >= 60:
        return "B"
    elif m >= 40:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(assign_grade)
print(df)
```
