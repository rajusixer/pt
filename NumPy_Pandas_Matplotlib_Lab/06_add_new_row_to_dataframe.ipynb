```python
import pandas as pd
df = pd.DataFrame({"Name": ["Amit", "Riya", "Sahil"],
                   "Department": ["CS", "IT", "CS"],
                   "Marks": [78, 85, 66]})
print("Original DataFrame:")
print(df)
new_row = pd.DataFrame({"Name": ["Neha"], "Department": ["IT"], "Marks": [91]})
df = pd.concat([df, new_row], ignore_index=True)
print("\nDataFrame after adding a new row:")
print(df)
```
