```python
import pandas as pd
import numpy as np
df = pd.DataFrame({"Name": ["Amit", "Riya", "Sahil", "Neha"],
                   "Department": ["CS", None, "CS", "IT"],
                   "Marks": [78, 85, np.nan, 91]})
print("DataFrame:")
print(df)
print("\nMissing values (True = missing):")
print(df.isnull())
print("\nCount of missing values per column:")
print(df.isnull().sum())
```
