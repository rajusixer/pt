```python
import pandas as pd
df = pd.DataFrame({"Name": ["Amit", "Riya", "Sahil", "Neha", "Karan", "Pooja"],
                   "Department": ["CS", "IT", "CS", "IT", "ME", "ME"],
                   "Marks": [78, 85, 66, 91, 72, 80]})
print("Student records:")
print(df)
avg = df.groupby("Department")["Marks"].mean()
print("\nAverage marks by department:")
print(avg)
```
