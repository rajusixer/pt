```python
import pandas as pd

def assign_grade(marks):
    if marks >= 85:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "D"

df = pd.DataFrame({"Name": ["Amit", "Riya", "Sahil", "Neha", "Karan"],
                   "Marks": [78, 85, 45, 91, 62]})
df["Grade"] = df["Marks"].apply(assign_grade)
print(df)
```
