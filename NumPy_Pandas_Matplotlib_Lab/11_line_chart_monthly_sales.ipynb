```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 250, 300, 280, 350]
plt.plot(months, sales, marker="o", color="black")
plt.title("Monthly Sales of a Product")
plt.xlabel("Month")
plt.ylabel("Sales (units)")
plt.grid(True)
plt.show()
```
