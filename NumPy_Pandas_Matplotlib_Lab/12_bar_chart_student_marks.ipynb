```python
import matplotlib.pyplot as plt

students = ["Amit", "Riya", "Sahil", "Neha", "Karan"]
marks = [78, 85, 66, 91, 72]
plt.bar(students, marks, color="steelblue")
plt.title("Marks of Five Students")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```
